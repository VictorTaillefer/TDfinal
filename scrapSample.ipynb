```python
import pandas as pd
import feedparser
import requests
import re
import json
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import datetime
import time
```

```python
urls = {
    "Alertes": "https://www.cert.ssi.gouv.fr/alerte/feed/",
    "Avis": "https://www.cert.ssi.gouv.fr/avis/feed/"
    
}

bulletins = []

for type_bulletin, url in urls.items():
    rss_feed = feedparser.parse(url)
    for entry in rss_feed.entries:
        bulletins.append({
            "id_bulletin": entry.link.split('/')[-2],
            "titre": entry.title,
            "type": type_bulletin,
            "date_publication": entry.published,
            "lien": entry.link
        })

print(f"{len(bulletins)} bulletins extraits avec succès.")
print("Exemple de bulletin :", bulletins[0])
```

```python
cve_id = "CVE-2023-24488"
url = f"https://cveawg.mitre.org/api/cve/{cve_id}"
response = requests.get(url)
data = response.json()
# Extraire la description
description = data["containers"]["cna"]["descriptions"][0]["value"]
# Extraire le score CVSS
#ATTENTION tous les CVE ne contiennent pas nécessairement ce champ, gérez l’exception,
#ou peut etre au lieu de cvssV3_0 c’est cvssV3_1 ou autre clé
cvss_score =data["containers"]["cna"]["metrics"][0]["cvssV3_1"]["baseScore"]
cwe = "Non disponible"
cwe_desc="Non disponible"
problemtype = data["containers"]["cna"].get("problemTypes", {})
if problemtype and "descriptions" in problemtype[0]:
    cwe = problemtype[0]["descriptions"][0].get("cweId", "Non disponible")
    cwe_desc=problemtype[0]["descriptions"][0].get("description", "Non disponible")
# Extraire les produits affectés
affected = data["containers"]["cna"]["affected"]
for product in affected:
    vendor = product["vendor"]
    product_name = product["product"]
    versions = [v["version"] for v in product["versions"] if v["status"] == "affected"]
    print(f"Éditeur : {vendor}, Produit : {product_name}, Versions : {', '.join(versions)}")
# Afficher les résultats
print(f"CVE : {cve_id}")
print(f"Description : {description}")
print(f"Score CVSS : {cvss_score}")
print(f"Type CWE : {cwe}")
print(f"CWE Description : {cwe_desc}")
```

```python
def get_severity(score):
    """Retourne la criticité basée sur le score CVSS."""
    try:
        score = float(score)
        if score == 0.0: return "None"
        elif score < 4.0: return "Low"
        elif score < 7.0: return "Medium"
        elif score < 9.0: return "High"
        else: return "Critical"
    except:
        return "Non disponible"

def get_mitre_details(cve_id):
    """Récupère les infos depuis l'API MITRE avec une gestion robuste des clés manquantes."""
    url = f"https://cveawg.mitre.org/api/cve/{cve_id}"
    details = {
        "cvss": "Non disponible",
        "severity": "Non disponible",
        "cwe": "Non disponible",
        "description": "Non disponible",
        "vendor": "Non disponible",
        "product": "Non disponible",
        "versions": "Non disponible"
    }
    
    try:
        response = requests.get(url, timeout=10)
        if response.status_code == 200:
            data = response.json()
            cna = data.get("containers", {}).get("cna", {})
            
            try:
                details["description"] = cna.get("descriptions", [{}])[0].get("value", "Non disponible")
            except: pass
            
            try:
                metrics = cna.get("metrics", [])
                for m in metrics:
                    if "cvssV3_1" in m:
                        details["cvss"] = m["cvssV3_1"].get("baseScore", "Non disponible")
                        break
                    elif "cvssV3_0" in m:
                        details["cvss"] = m["cvssV3_0"].get("baseScore", "Non disponible")
                        break
                if details["cvss"] != "Non disponible":
                    details["severity"] = get_severity(details["cvss"])
            except: pass
            
            try:
                problem_types = cna.get("problemTypes", [])
                if problem_types:
                    details["cwe"] = problem_types[0].get("descriptions", [{}])[0].get("cweId", "Non disponible")
            except: pass
            
            try:
                affected = cna.get("affected", [])
                if affected:
                    details["vendor"] = affected[0].get("vendor", "Non disponible")
                    details["product"] = affected[0].get("product", "Non disponible")
                    
                    versions_affectees = []
                    for v in affected[0].get("versions", []):
                        if v.get("status") == "affected":
                            versions_affectees.append(v.get("version", ""))
                    if versions_affectees:
                        details["versions"] = ", ".join(versions_affectees)
            except: pass
            
    except Exception as e:
        print(f"Erreur MITRE pour {cve_id}: {e}")
        
    return details

def get_epss_score(cve_id):
    """Récupère le score EPSS depuis l'API FIRST."""
    url = f"https://api.first.org/data/v1/epss?cve={cve_id}"
    try:
        response = requests.get(url, timeout=10)
        if response.status_code == 200:
            data = response.json()
            epss_data = data.get("data", [])
            if epss_data:
                return epss_data[0].get("epss", "Non disponible")
    except Exception as e:
        print(f"Erreur EPSS pour {cve_id}: {e}")
    return "Non disponible"
```

```python
donnees_enrichies = []

print(f"Début de l'enrichissement pour {len(bulletins_list)} CVEs...")
i=0
for bulletin in bulletins_list[:30]:
    i+=1
    cve_id = bulletin['cve_id']
    print(f"Traitement de {cve_id}... {i}/30")
    
    # Appels API
    mitre_data = get_mitre_details(cve_id)
    epss_score = get_epss_score(cve_id)
    
    ligne = {
        "ID ANSSI": bulletin.get("id_bulletin"),
        "Titre ANSSI": bulletin.get("titre"),
        "Type": bulletin.get("type"),
        "Date": bulletin.get("date_publication"),
        "CVE": cve_id,
        "CVSS": mitre_data["cvss"],
        "Base Severity": mitre_data["severity"],
        "CWE": mitre_data["cwe"],
        "EPSS": epss_score,
        "Lien": bulletin.get("lien"),
        "Description": mitre_data["description"],
        "Éditeur": mitre_data["vendor"],
        "Produit": mitre_data["product"],
        "Versions affectées": mitre_data["versions"]
    }
    donnees_enrichies.append(ligne)
    time.sleep(2)
    

print("Enrichissement terminé lets gooo la teaaaam !")
```

```python
import pandas as pd

df_cve = pd.DataFrame(donnees_enrichies)

df_cve['CVSS'] = pd.to_numeric(df_cve['CVSS'], errors='coerce')
df_cve['EPSS'] = pd.to_numeric(df_cve['EPSS'], errors='coerce')
df_cve['Date'] = pd.to_datetime(df_cve['Date'], format='mixed', errors='coerce')

display(df_cve.head())

df_cve.to_csv("sample_data_from_scrap.csv", index=False)
print("Données consolidées et sauvegardées dans 'sample_data_from_scrap.csv'")
```
